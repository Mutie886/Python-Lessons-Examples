### Exercise 31:
Write a Python script that takes a list as input and returns a new list with all duplicates removed.

```python
# Exercise 31
```

---

````python
l_nums = int(input("Enter how many numbers you want to add: "))
my_list = []

for i in range(l_nums):
    number = input(f"Enter number {i+1}: ")
    my_list.append(number)

print("Your list is:", my_list)
````
