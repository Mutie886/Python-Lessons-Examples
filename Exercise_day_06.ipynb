#Exercise

write a program which computes the following equation: $ ax^2 + bx + c = 0 $ when $x = 3$, $a != 0$, $b != 0$, $c != 0$

```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))
c = int(input("Enter c: "))

delta = b * b - 4 * a * c #print delta
if a != 0:
  if delta > 0:
    print("We have two solutions ", (- b - delta**0.5)/2 * a)
  elif delta == 0 :
    print("We have double root: ", -b/(2*a))
  else:
    print("We dont hv reals solutions")

else:
  print("It is not second degree equation")
```

##Loops

initialize increment variable

while condition:

      instructions block

      increment

```python
i = 1
while (i < 10):
  print("The value of i is ", i)
  i +=1
```

```python
#use loop with break
# Stop when number reaches 3
i = 1
while True:
    print(i)
    if i == 3:
        break
    i += 1
```

```python
##use loop with continue
# Skip even numbers
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
```

```python
# Keep asking until user types 'exit'
my_input = ""
while my_input != "exit":
    my_input = input("Type 'exit' to stop: ")
```

```python
##countdown loop
# Countdown from 7 to 1
n = 7
while n > 0:
    print(n)
    n -= 1
```
